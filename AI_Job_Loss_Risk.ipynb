#### Data Cleaning

#### Check if the are null or NaN rows

```python
df.isnull().sum()
```

```python
# Fill missing values with the median of the worker's industry (the target is NOT used)
for col in ['Digital_Skills_Score', 'Monthly_Income_ZAR', 'Industry_AI_Adoption']:
    df[col] = df.groupby('Industry')[col].transform(lambda x: x.fillna(x.median()))

print(df.isnull().sum())
```

#### Check duplicates

```python
df.duplicated().sum()
```

```python
df_before_duplicates = df.copy()

print("Before removing duplicates:", df.shape)
print("Rows involved in duplicates:", df.duplicated(keep=False).sum())
print("Rows that will be removed:", df.duplicated().sum())
```

```python
df = df.drop_duplicates().reset_index(drop=True)

print("After removing duplicates:", df.shape)
print("Remaining exact duplicates:", df.duplicated().sum())
```

```python
df.columns
```

```python
print(df["High_Risk"].value_counts())
```

```python
numerical_columns = [
    "Age",
    "Years_Experience",
    "Digital_Skills_Score",
    "Monthly_Income_ZAR",
    "Routine_Task_Index",
    "Industry_AI_Adoption",
    "Automation_Exposure_Index"
]

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print("Column:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print("Number of outliers:", len(outliers))
    print("-" * 50)
```

#### Handle the income outliers (cap at the IQR upper limit)

```python
# Income is right-skewed and has a few extreme values (6-12x too high).
# We cap them instead of deleting the rows so we do not lose workers.
Q1 = df["Monthly_Income_ZAR"].quantile(0.25)
Q3 = df["Monthly_Income_ZAR"].quantile(0.75)
upper_limit = Q3 + 1.5 * (Q3 - Q1)

print("Income cap:", round(upper_limit))
print("Rows capped:", (df["Monthly_Income_ZAR"] > upper_limit).sum())

df["Monthly_Income_ZAR"] = df["Monthly_Income_ZAR"].clip(upper=upper_limit)
```

#### Prepare the data

```python
# Categorical columns must be turned into numbers (one-hot encoding)
categorical_columns = [
    "Province",
    "Metro_Status",
    "Industry",
    "Gender",
    "Education",
    "Skill_Level",
    "Employment_Type",
    "Union_Member",
    "Firm_Size",
    "Recent_Training"
]

numeric_features = [
    "Age",
    "Years_Experience",
    "Digital_Skills_Score",
    "Monthly_Income_ZAR",
    "Routine_Task_Index",
    "Industry_AI_Adoption",
    "Automation_Exposure_Index"
]

# Features and target
# (Year is kept only for dashboard filtering, it is not used as a feature)
X = pd.get_dummies(
    df[numeric_features + categorical_columns],
    columns=categorical_columns,
    dtype=int
)

y = df["High_Risk"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

####  Train/Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target:")
print(y_train.value_counts())

print("\nTesting target:")
print(y_test.value_counts())
```

#### Balance ONLY the training data

```python
print(y_train.value_counts())
```

#### Balance the training data manually

```python
# Combine X_train and y_train
train_data = X_train.copy()
train_data["High_Risk"] = y_train.values

# Separate the two classes
low_risk = train_data[train_data["High_Risk"] == 0]
high_risk = train_data[train_data["High_Risk"] == 1]

# Find the larger class
max_count = max(len(low_risk), len(high_risk))

# Oversample the smaller class
low_risk_oversampled = low_risk.sample(
    n=max_count,
    replace=True,
    random_state=42
)

high_risk_oversampled = high_risk.sample(
    n=max_count,
    replace=True,
    random_state=42
)

# Combine the classes
train_balanced = pd.concat(
    [low_risk_oversampled, high_risk_oversampled]
)

# Shuffle the data
train_balanced = train_balanced.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

# Separate X and y
X_train_balanced = train_balanced.drop(
    "High_Risk",
    axis=1
)
y_train_balanced = train_balanced["High_Risk"]

# Check the results
print("Before balancing:")
print(y_train.value_counts())

print("\nAfter oversampling:")
print(y_train_balanced.value_counts())
```

```python
print("Test set:")
print(y_test.value_counts())
```

#### Train the ML models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.svm import SVC
```

#### Create the models

```python

models_tuned = {

    "Logistic Regression": LogisticRegression(
        max_iter=2000
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=7,
        min_samples_split=20,
        min_samples_leaf=10,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_depth=8,
        min_samples_split=20,
        min_samples_leaf=10,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=150,
        max_depth=3,
        min_samples_split=20,
        min_samples_leaf=10,
        learning_rate=0.05,
        random_state=42
    )
}
```

#### Train and test all models

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

results = []

for name, model in models_tuned.items():

    print(f"\nTraining {name}...")

    # Train on balanced training data
    model.fit(
        X_train_balanced,
        y_train_balanced
    )

    print(f"Testing {name}...")

    # Training accuracy
    train_pred = model.predict(X_train_balanced)
    train_accuracy = accuracy_score(
        y_train_balanced,
        train_pred
    )

    # Testing accuracy
    test_pred = model.predict(X_test)
    test_accuracy = accuracy_score(
        y_test,
        test_pred
    )

    difference = train_accuracy - test_accuracy

    results.append({
        "Model": name,
        "Training Accuracy": train_accuracy * 100,
        "Testing Accuracy": test_accuracy * 100,
        "Difference": difference * 100,
        "Precision": precision_score(y_test, test_pred) * 100,
        "Recall": recall_score(y_test, test_pred) * 100,
        "F1": f1_score(y_test, test_pred) * 100
    })

    print(f"Training Accuracy: {train_accuracy * 100:.2f}%")
    print(f"Testing Accuracy: {test_accuracy * 100:.2f}%")
    print(f"Difference: {difference * 100:.2f}%")
```

```python
results_df = pd.DataFrame(results)

print("\n===== TUNED MODEL RESULTS =====")
print(results_df.round(2))
```

#### Interpreting the trade-off (precision vs recall)

* **Recall** (of the High Risk class) = of all workers who really are at high risk, how many did the model catch?
* **Precision** = of all workers the model flagged, how many really are at high risk?

For an *early-intervention / re-skilling programme* a missed worker (false negative) is more costly than an extra worker offered training (false positive), so **recall is preferred**. For *scarce, expensive support* (e.g. retrenchment packages) precision matters more.

#### ROC_AUC (ROC = Receiver Operating Characteristic, AUC = Area Under the Curve)

```python
from sklearn.metrics import roc_auc_score

# Get probability of class 1 (HIGH RISK)
y_prob = random_forest.predict_proba(X_test)[:, 1]

# Calculate ROC-AUC
roc_auc = roc_auc_score(y_test, y_prob)

print("Random Forest ROC-AUC:", roc_auc)
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Calculate ROC curve
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

# Calculate ROC-AUC
auc = roc_auc_score(y_test, y_prob)

# Plot ROC curve
plt.plot(fpr, tpr, label=f"Random Forest ROC-AUC = {auc:.3f}")

# Random guessing line
plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - AI Job Loss Risk Prediction")
plt.legend()
plt.show()
```

#### Choosing a threshold for early intervention (recall vs precision)

```python
from sklearn.metrics import precision_score, recall_score

for threshold in [0.30, 0.40, 0.50, 0.60]:
    pred = (y_prob >= threshold).astype(int)
    print(
        f"Threshold {threshold:.2f} -> "
        f"Precision: {precision_score(y_test, pred):.2f} | "
        f"Recall: {recall_score(y_test, pred):.2f}"
    )
```
